## **Image Pixel Adjustment**

### Using Linear Equation
$$I_{new} = I \cdot a + b$$

Please perform at ***least 20 iterations*** with varying values of $a$ and $b$.
Assign the result of each iteration to an array.<br/>

```python
### START CODE HERE ###

a = 0
b = 0

lin_result = []

for i in range(40):
    adjusted_img = rgb_img.astype(np.float32) * a + b
    adjusted_img = np.clip(adjusted_img, 0, 255).astype(np.uint8)
    lin_result.append(adjusted_img)
    a += 0.1
    b = i * 3

### END CODE HERE ###
```

Save the array of images to a video. Please complete the function below.

```python
### START CODE HERE ###
def write_images_to_video(image_array, output_file, frame_rate=30):
    (h, w) = image_array[0].shape[:2]

    fourcc = cv2.VideoWriter_fourcc(*"MP4V")
    out = cv2.VideoWriter(output_file, fourcc, frame_rate, (h, w))
    
    for img in image_array:
        out.write(img)
        
### END CODE HERE ###
```

### Gamma Equation
$$I_{new} = I^{\gamma} \cdot a + b$$

Where $a$ is 1 and $b$ is 0. 
Please perform ***at least 10 iterations*** for 0 < $\gamma$ < 1 and another ***at least 10 iterations*** for $\gamma$ > 1
Assign the result of each iteration to an array.<br/>


```python
### START CODE HERE ###
a = 1
b = 0
gamma = 0.1

gam_result = []

for i in range(20):
    normalized = rgb_img.astype(np.float32) / 255.0
    
    corrected = (normalized ** gamma) * 255
    
    gamma_img = np.clip(corrected, 0, 255).astype(np.uint8)
    gam_result.append(gamma_img)
    gamma += 0.1

### END CODE HERE ###
```

## **Questions**
1. If an image having over brightness or saturation, what would be possible solutions to enhance the image.
2. How can we make an image reverse intensity (black-to-white, white-to-black)? Explain the process

1. over brightness สามารถแก้ไขได้โดยใช้ linear equation โดยใช้ค่า a < 1 และ ค่า b น้อย ๆ จะทำให้ความเข้มของสีในรูปลดลง ส่งผลให้ความสว่างของรูปลดลง
   over saturation สามารถแก้ไขได้โดยใช้ gamma equation โดยใช้ค่า gamma ในช่วง (0, 1) จะได้ภาพที่ saturation ลดลง

```python
over_brightness = lin_result[22]
bright_enhanced = ((over_brightness.astype(np.float32) * 0.3)).astype(np.uint8)

over_saturation = gam_result[19]
sat_enhanced = (((over_saturation.astype(np.float32) / 255) ** 0.5) * 255).astype(np.uint8)

fig = plt.figure(figsize=(8, 6))
grid = ImageGrid(fig, rect=111 , nrows_ncols=(2, 2))

titles = ("Over Brightness", "Brightness Enhanced", "Over Saturation", "Saturation Enhanced")
images = (over_brightness, bright_enhanced, over_saturation, sat_enhanced)

for ax, title, image in zip(grid, titles, images):
    ax.imshow(image)
    ax.set_title(title)
    ax.set_axis_off()
        
plt.show()
```

2. ใช้วิธี linear equation โดยกำหนดค่า a เป็น -1

```python
reversed_img = ((rgb_img.astype(np.float32) * -1)).astype(np.uint8)

fig = plt.figure(figsize=(12, 8))
grid = ImageGrid(fig, rect=111 , nrows_ncols=(1, 2))

titles = ("Original", "Reversed")
images = (rgb_img, reversed_img)

for ax, title, image in zip(grid, titles, images):
    ax.imshow(image)
    ax.set_title(title)
    ax.set_axis_off()
        
plt.show()
```
